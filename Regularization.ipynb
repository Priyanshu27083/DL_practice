```python

# keras imports for the dataset and building our neural network
from keras.datasets import mnist
from keras.models import Sequential
from keras.layers import Dense, Dropout, Conv2D, MaxPool2D, Flatten

from tensorflow.keras.utils import to_categorical



# to calculate accuracy
from sklearn.metrics import accuracy_score

# loading the dataset
(X_train, y_train), (X_test, y_test) = mnist.load_data()
#X_train and y_train are the training images and their labels, respectively, while X_test and y_test are the test images and labels.

# Reshapes the data so it has the shape (samples, height, width, channels).
X_train = X_train.reshape(X_train.shape[0], 28, 28, 1)
X_test = X_test.reshape(X_test.shape[0], 28, 28, 1)
X_train = X_train.astype('float32')
X_test = X_test.astype('float32')
#Converts the data type of the image data to float32 for better precision during training.



# #Divides the pixel values by 255 to normalize the images.
#This scales the values to the range [0, 1], which makes the training more stable and faster.
X_train /= 255
X_test /= 255

# one-hot encoding using keras' numpy-related utilities
n_classes = 10
print("Shape before one-hot encoding: ", y_train.shape)
Y_train = to_categorical(y_train, n_classes)
Y_test = to_categorical(y_test, n_classes)
print("Shape after one-hot encoding: ", Y_train.shape)

#The labels (y_train, y_test) are converted to one-hot encoded format.



# building a linear stack of layers with the sequential model
model = Sequential()
# convolutional layer

model.add(Conv2D(25, kernel_size=(3,3), strides=(1,1), padding='valid', activation='relu', input_shape=(28,28,1)))



model.add(MaxPool2D(pool_size=(1,1)))


# flatten output of conv

model.add(Flatten())




# hidden layer

model.add(Dense(100, activation='relu'))



# output layer
model.add(Dense(10, activation='softmax'))

# compiling the sequential model
model.compile(loss='categorical_crossentropy', metrics=['accuracy'], optimizer='adam')

# training the model for 10 epochs
model.fit(X_train, Y_train, batch_size=128, epochs=10, validation_data=(X_test, Y_test))
```

### Loading and Preprocessing GTSRB Dataset

```python
import zipfile
import os
import pandas as pd
import numpy as np
from PIL import Image
from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical
gtsrb_download_url = 'https://sid.erda.dk/public/archives/daaeac0d7ce1152aea9b61d9f1e19370/GTSRB_Final_Training_Images.zip'

# Path where the zip file will be saved
gtsrb_zip_path = '/content/GTSRB_Final_Training_Images.zip'
output_dir = '/content/' # Extract directly into /content to simplify subsequent paths

# Create output directory if it doesn't exist (already created by colab usually)
os.makedirs(output_dir, exist_ok=True)

# Download the GTSRB dataset if it doesn't exist
if not os.path.exists(gtsrb_zip_path):
    print(f"Downloading GTSRB dataset from {gtsrb_download_url}...")
    !wget -O {gtsrb_zip_path} {gtsrb_download_url}
    print("Download complete.")
else:
    print("GTSRB zip file already exists. Skipping download.")

print(f"Unzipping GTSRB dataset to: {output_dir}")
with zipfile.ZipFile(gtsrb_zip_path, 'r') as zip_ref:
    zip_ref.extractall(output_dir)

print("GTSRB dataset unzipped.")
```

```python
# Load the training data (images and labels)
# The training data is organized into folders 00000 through 00042, each representing a class.
# Inside each class folder, there's a CSV file (e.g., 'GT-00000.csv') with image information.

img_paths = []
labels = []

# Loop through all class folders (00000-00042)
for class_id in range(0, 43):
    # The extracted content usually has a top-level 'GTSRB' folder
    class_folder = os.path.join(output_dir, 'GTSRB', 'Final_Training', 'Images', f'{class_id:05d}')
    csv_file = os.path.join(class_folder, f'GT-{class_id:05d}.csv')

    if os.path.exists(csv_file):
        df_labels = pd.read_csv(csv_file, sep=';')
        for index, row in df_labels.iterrows():
            img_name = row['Filename']
            # For GTSRB, it's typically just the image file name directly in the class folder
            img_path = os.path.join(class_folder, img_name)
            if os.path.exists(img_path):
                img_paths.append(img_path)
                labels.append(row['ClassId'])

print(f"Found {len(img_paths)} images and {len(labels)} labels.")
```

```python
IMG_HEIGHT = 30
IMG_WIDTH = 30
NUM_CLASSES = 43

# Function to load and preprocess images
def load_and_preprocess_image(image_path):
    img = Image.open(image_path)
    img = img.resize((IMG_WIDTH, IMG_HEIGHT))
    img = np.array(img) / 255.0  # Normalize to [0, 1]
    return img

# Load all images
images = np.array([load_and_preprocess_image(p) for p in img_paths])
labels = np.array(labels)

# Convert labels to one-hot encoding
labels_one_hot = to_categorical(labels, NUM_CLASSES)

print(f"Images shape: {images.shape}")
print(f"Labels shape (one-hot): {labels_one_hot.shape}")

# Split data into training and validation sets
x_train_gtsrb, x_val_gtsrb, y_train_gtsrb, y_val_gtsrb = train_test_split(
    images, labels_one_hot, test_size=0.2, random_state=42, stratify=labels
)

print(f"GTSRB Training data shape: {x_train_gtsrb.shape}")
print(f"GTSRB Validation data shape: {x_val_gtsrb.shape}")
```

### L2 Regularization on GTSRB Dataset

```python
import tensorflow as tf
from tensorflow.keras import layers, models, regularizers

# Define a simple model with L2 regularization for GTSRB using only Dense layers
def build_gtsrb_l2_model():
    model = models.Sequential([
        layers.Flatten(input_shape=(IMG_HEIGHT, IMG_WIDTH, 3)), # Flatten the image data
        layers.Dense(128, activation='relu',
                     kernel_regularizer=regularizers.l2(0.001)),
        layers.Dense(64, activation='relu',
                     kernel_regularizer=regularizers.l2(0.001)),
        layers.Dense(NUM_CLASSES, activation='softmax')
    ])
    return model

model_gtsrb_l2 = build_gtsrb_l2_model()
model_gtsrb_l2.summary()
```

```python
# Compile the L2 regularized model
model_gtsrb_l2.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("Model compiled successfully.")
```

```python
# Train the L2 regularized model on GTSRB data
history_gtsrb_l2 = model_gtsrb_l2.fit(
    x_train_gtsrb,
    y_train_gtsrb,
    epochs=10, # You can adjust the number of epochs
    validation_data=(x_val_gtsrb, y_val_gtsrb),
    batch_size=32
)

print("Model training complete.")
```

```python
# Evaluate the L2 regularized model
loss_gtsrb_l2, accuracy_gtsrb_l2 = model_gtsrb_l2.evaluate(x_val_gtsrb, y_val_gtsrb)

print(f"Test Loss with L2 regularization: {loss_gtsrb_l2:.4f}")
print(f"Test Accuracy with L2 regularization: {accuracy_gtsrb_l2:.4f}")
```

### Early Stopping on GTSRB Dataset

```python
import tensorflow as tf
from tensorflow.keras.utils import to_categorical
from tensorflow.keras import layers, models
from tensorflow.keras.callbacks import EarlyStopping

# GTSRB dataset (x_train_gtsrb, y_train_gtsrb, x_val_gtsrb, y_val_gtsrb) is already loaded and preprocessed.
# IMG_HEIGHT, IMG_WIDTH, NUM_CLASSES are also defined from the GTSRB loading.

# The GTSRB data is already normalized and one-hot encoded.
# x_train_gtsrb shape: (samples, IMG_HEIGHT, IMG_WIDTH, 3)
# y_train_gtsrb shape: (samples, NUM_CLASSES)

print(f"GTSRB Training data shape: {x_train_gtsrb.shape}")
print(f"GTSRB Validation data shape: {x_val_gtsrb.shape}")
```

```python
# Build simple model for GTSRB (NO L2 regularization, using Flatten and Dense layers)
def build_gtsrb_early_stopping_model():
    model = models.Sequential([
        layers.Flatten(input_shape=(IMG_HEIGHT, IMG_WIDTH, 3)),
        layers.Dense(64, activation='relu'),
        layers.Dense(64, activation='relu'),
        layers.Dense(NUM_CLASSES, activation='softmax')
    ])
    return model

model_gtsrb_es = build_gtsrb_early_stopping_model()
model_gtsrb_es.summary()
```

```python
model_gtsrb_es.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("GTSRB model for Early Stopping compiled successfully.")
```

```python
# Early Stopping callback
earlystop = EarlyStopping(
    monitor='val_accuracy',   # Monitor validation accuracy
    patience=5,               # Stop if no improvement after 5 epochs (adjusted from 3 to 5 for potential benefit)
    restore_best_weights=True # Restore best model weights
)

epochs = 25  # Increased epochs to allow Early Stopping more room to work
batch_size = 32   # Consistent batch size with previous GTSRB training

# Train model with Early Stopping on GTSRB data
history_gtsrb_es = model_gtsrb_es.fit(
    x_train_gtsrb,
    y_train_gtsrb,
    epochs=epochs,
    batch_size=batch_size,
    validation_data=(x_val_gtsrb, y_val_gtsrb), # Use already split validation data
    callbacks=[earlystop]
)

print("GTSRB model training with Early Stopping complete.")
```

```python
# Evaluate model
test_loss_gtsrb_es, test_acc_gtsrb_es = model_gtsrb_es.evaluate(x_val_gtsrb, y_val_gtsrb)

print(f"Test Loss with Early Stopping on GTSRB: {test_loss_gtsrb_es:.4f}")
print(f"Test Accuracy with Early Stopping on GTSRB: {test_acc_gtsrb_es:.4f}")
```

### Adding Noise to Inputs and Outputs on GTSRB Dataset

```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.utils import to_categorical
from tensorflow.keras import layers, models

# GTSRB dataset (x_train_gtsrb, y_train_gtsrb, x_val_gtsrb, y_val_gtsrb) is already loaded and preprocessed.
# IMG_HEIGHT, IMG_WIDTH, NUM_CLASSES are also defined from the GTSRB loading.

print(f"Original GTSRB Training data shape: {x_train_gtsrb.shape}")
print(f"Original GTSRB Validation data shape: {x_val_gtsrb.shape}")
print(f"Original GTSRB Training labels shape: {y_train_gtsrb.shape}")
```

```python
# Flatten GTSRB data for Dense layers
# Each image is 30x30x3, so flattened size is 30*30*3 = 2700
input_dim_gtsrb_flat = IMG_HEIGHT * IMG_WIDTH * 3
x_train_gtsrb_flat = x_train_gtsrb.reshape(-1, input_dim_gtsrb_flat)
x_val_gtsrb_flat = x_val_gtsrb.reshape(-1, input_dim_gtsrb_flat)

# ===============================
# 🔹 Add Noise to INPUTS
# ===============================
noise_factor = 0.2
x_train_noisy_gtsrb = x_train_gtsrb_flat + noise_factor * np.random.normal(
    loc=0.0, scale=1.0, size=x_train_gtsrb_flat.shape)

# Keep pixel values between 0 and 1
x_train_noisy_gtsrb = np.clip(x_train_noisy_gtsrb, 0., 1.)

print(f"Noisy GTSRB Training inputs shape: {x_train_noisy_gtsrb.shape}")
```

```python
# ===============================
# 🔹 Add Noise to OUTPUTS (Label Noise)
# Randomly flip 10% of labels
# ===============================
noise_ratio = 0.1
num_samples_to_noise = int(noise_ratio * y_train_gtsrb.shape[0])

# Create a copy of y_train_gtsrb to add noise to, leaving original clean
y_train_noisy_gtsrb = y_train_gtsrb.copy()

random_indices = np.random.choice(y_train_gtsrb.shape[0], num_samples_to_noise, replace=False)

for idx in random_indices:
    # Choose a random class ID different from the current one
    current_class = np.argmax(y_train_noisy_gtsrb[idx])
    possible_random_classes = list(range(NUM_CLASSES))
    possible_random_classes.remove(current_class) # Ensure it's a different class
    random_label_id = np.random.choice(possible_random_classes)

    # Set the new one-hot encoded label
    y_train_noisy_gtsrb[idx] = to_categorical(random_label_id, NUM_CLASSES)

print(f"Noisy GTSRB Training labels shape: {y_train_noisy_gtsrb.shape}")
```

```python
# ===============================
# Build Model (same as original, adapted for GTSRB flat input)
# ===============================
def build_gtsrb_noisy_model():
    model = models.Sequential([
        layers.Dense(128, activation='relu', input_shape=(input_dim_gtsrb_flat,)),
        layers.Dense(64, activation='relu'),
        layers.Dense(NUM_CLASSES, activation='softmax')
    ])
    return model

model_gtsrb_noisy = build_gtsrb_noisy_model()
model_gtsrb_noisy.summary()
```

```python
model_gtsrb_noisy.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

print("GTSRB model with noise compiled successfully.")
```

```python
epochs = 20
batch_size = 256

# Train model using noisy inputs and noisy labels, validate on clean data
history_gtsrb_noisy = model_gtsrb_noisy.fit(
    x_train_noisy_gtsrb,
    y_train_noisy_gtsrb,
    epochs=epochs,
    batch_size=batch_size,
    validation_data=(x_val_gtsrb_flat, y_val_gtsrb) # Validate on clean data
)

print("GTSRB model training with noise complete.")
```

```python
# Evaluate on clean validation data
test_loss_gtsrb_noisy, test_acc_gtsrb_noisy = model_gtsrb_noisy.evaluate(x_val_gtsrb_flat, y_val_gtsrb)

print(f"Test Loss with Input and Label Noise on GTSRB: {test_loss_gtsrb_noisy:.4f}")
print(f"Test Accuracy with Input and Label Noise on GTSRB: {test_acc_gtsrb_noisy:.4f}")
```

### Parameter Sharing on GTSRB Dataset

```python
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense, Concatenate
from tensorflow.keras.models import Model

# GTSRB data is already loaded and preprocessed:
# x_train_gtsrb_flat, y_train_gtsrb
# x_val_gtsrb_flat, y_val_gtsrb
# input_dim_gtsrb_flat, NUM_CLASSES

print(f"GTSRB Flattened Training data shape: {x_train_gtsrb_flat.shape}")
print(f"GTSRB Flattened Validation data shape: {x_val_gtsrb_flat.shape}")
print(f"GTSRB Training labels shape: {y_train_gtsrb.shape}")

# 🔹 Shared Layer
shared_dense_gtsrb = Dense(64, activation='relu', name='gtsrb_shared_dense')

# Two inputs, adapted for GTSRB flattened input dimension
input1_gtsrb = Input(shape=(input_dim_gtsrb_flat,), name='gtsrb_input_1')
input2_gtsrb = Input(shape=(input_dim_gtsrb_flat,), name='gtsrb_input_2')

# Apply SAME layer (parameter sharing)
encoded1_gtsrb = shared_dense_gtsrb(input1_gtsrb)
encoded2_gtsrb = shared_dense_gtsrb(input2_gtsrb)

# Merge
merged_gtsrb = Concatenate()([encoded1_gtsrb, encoded2_gtsrb])
output_gtsrb = Dense(NUM_CLASSES, activation='softmax', name='gtsrb_output')(merged_gtsrb)

model_gtsrb_ps = Model(inputs=[input1_gtsrb, input2_gtsrb], outputs=output_gtsrb)

model_gtsrb_ps.summary()
```

```python
model_gtsrb_ps.compile(optimizer='adam',
                       loss='categorical_crossentropy',
                       metrics=['accuracy'])

print("GTSRB Parameter Sharing model compiled successfully.")
```

```python
epochs_ps = 10 # You can adjust the number of epochs
batch_size_ps = 256 # Consistent batch size

# Train (using x_train_gtsrb_flat twice for demo, and y_train_gtsrb for labels)
history_gtsrb_ps = model_gtsrb_ps.fit([x_train_gtsrb_flat, x_train_gtsrb_flat], y_train_gtsrb,
                                      epochs=epochs_ps,
                                      batch_size=batch_size_ps,
                                      validation_data=([x_val_gtsrb_flat, x_val_gtsrb_flat], y_val_gtsrb))

print("GTSRB Parameter Sharing model training complete.")
```

```python
# Evaluate the Parameter Sharing model for GTSRB
test_loss_gtsrb_ps, test_acc_gtsrb_ps = model_gtsrb_ps.evaluate([x_val_gtsrb_flat, x_val_gtsrb_flat], y_val_gtsrb)

print(f"Test Loss with Parameter Sharing on GTSRB: {test_loss_gtsrb_ps:.4f}")
print(f"Test Accuracy with Parameter Sharing on GTSRB: {test_acc_gtsrb_ps:.4f}")
```
